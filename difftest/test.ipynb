```python
from zeusandatlas.differential import diff

t = [0, 0.1, 0.3, 0.4, 0.55, 0.67, 0.71]
x = [23.1, 22.5, 23.5, 21.88, 22.5, 23.5, 24.88]

v = diff(t, x)
print(v)
```

```python
import matplotlib.pyplot as plt

plt.figure()

plt.plot(t, x, marker='o')
plt.xlabel('t')
plt.ylabel('x')
plt.title('Signal x vs. t')
plt.savefig('x_vs_t.pdf')
plt.show()
```

```python
plt.figure()

plt.plot(t[1:], v, marker='o')
plt.xlabel('t')
plt.ylabel('v')
plt.title('Discrete Derivative v vs. t')
plt.savefig('v_vs_t.pdf')
plt.show()
```
